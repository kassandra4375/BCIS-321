```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid", palette="magma")

plt.rcParams['figure.dpi'] = 120
```

```python
df = pd.read_csv("winequality-red.csv", sep=';')

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Info:")
print(df.info())

print("\nSummary Statistics:")
print(df.describe())
```

```python
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df=df.drop_duplicates()
```

```python
correlation_matrix = df.corr()

print(correlation_matrix)
```

```python
#Visualization 1 - Histogram

plt.figure(figsize=(8,5))

plt.hist(df['quality'],
         bins=6,
         color='#5D001E',
         edgecolor='white')

plt.title("Distribution of Wine Quality Ratings",
          fontsize=18,
          fontweight='bold')

plt.xlabel("Quality Rating")
plt.ylabel("Frequency")

sns.despine()
plt.tight_layout()
plt.show()
```

```python
#Visualization 2 - Scatterplot
plt.figure(figsize=(8,5))

plt.scatter(df['alcohol'],
            df['quality'],
            c=df['quality'],
            cmap='magma',
            alpha=0.7)

plt.title("Alcohol Content vs Wine Quality",
          fontsize=18,
          fontweight='bold')

plt.xlabel("Alcohol", fontsize=12)
plt.ylabel("Quality", fontsize=12)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
#Visualization 3 - Heatmap
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(),
            annot=True,
            cmap='magma' ,
            linewidths=0.5)

plt.title("Correlation Heatmap of Wine Features",
          fontsize=20,
          fontweight='bold')

sns.despine()
plt.tight_layout()
plt.show()
```

```python
#Visualization 4 - Boxplot
plt.figure(figsize=(8,5))

sns.boxplot(x='quality',
            y='alcohol',
            data=df,
            palette='magma')

plt.title("Alcohol Content by Wine Quality",
          fontsize=18,
          fontweight='bold')

plt.xlabel("Wine Quality", fontsize=12)
plt.ylabel("Alcohol", fontsize=12)

plt.tight_layout()
sns.despine()
plt.show()
```

```python
#Visualization 5 - Volatile Acidity Boxplot
plt.figure(figsize=(8,5))

sns.boxplot(x='quality',
            y='volatile acidity',
            data=df,
            palette='magma')

plt.title("Volatile Acidity by Wine Quality",
          fontsize=18,
          fontweight='bold')
plt.xlabel("Quality", fontsize=12)
plt.ylabel("Volatile Acidity", fontsize=12)

sns.despine()
plt.show()
```
